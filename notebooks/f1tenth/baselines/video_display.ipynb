# Atlas-DeePC: F1TENTH Lap Video

A replay-verified **collision-free full lap** from the retained paper experiment:
**K=12, d=40, seed 542, σ=0.001**, cross-track RMSE **1.74 cm**.
This is the paper’s selected example, replayed from its original saved trajectory.

The dashboard shows the official map/reference, the saved true trajectory,
a moving close-up with the **0.58 × 0.31 m car footprint**, speed tracking, and
signed cross-track error. The orange dotted horizon is the saved active reference.
The overview circle is only a locator; judge track clearance from the close-up.

Playback is real time. The first 1.5 s is shared initialization, followed by the
22.1 s controller lap and a 1 s final hold. Warmup is included in collision
verification but excluded from the reported tracking RMSE. Pose interpolation is
for display only; errors and success use the saved records. Projection selection
stays on chart 0 in this example.

Here σ denotes the noise/signal **standard-deviation ratio**: 0.1% in amplitude,
or 0.0001% as a variance ratio. Both offline and online data use this noise model.

**Run All** displays the existing MP4, or renders it if missing. Viewing the saved
video does not require the experiment SSD; regeneration requires the saved trace
on the SSD. No controller solves or experiment reruns are launched.

```python
from pathlib import Path
import json
import sys
from IPython.display import Video, display

ROOT = next(root for parent in (Path.cwd(), *Path.cwd().parents)
            for root in (parent, parent / "code")
            if (root / "src/atlas_deepc").is_dir())
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))

RUN_DIR = (ROOT / "experiments/f1tenth/paper_nsr_20261027/sigma_0p001"
           / "atlas_k12_d40/seed542")
VIDEO_PATH = ROOT / "experiments/f1tenth/video_gallery_v1/atlas_seed542_lap.mp4"
RENDER_IF_MISSING = True
OVERWRITE = False
PLAYBACK_SPEED = 1.0  # Set OVERWRITE=True to regenerate at a different speed.
FPS = 30

if OVERWRITE or (RENDER_IF_MISSING and not VIDEO_PATH.is_file()):
    from atlas_deepc.f1tenth.lap_video import render_lap
    render_lap(RUN_DIR, VIDEO_PATH, fps=FPS,
               playback_speed=PLAYBACK_SPEED, overwrite=OVERWRITE)
if not VIDEO_PATH.is_file():
    raise FileNotFoundError(f"Missing {VIDEO_PATH}; enable RENDER_IF_MISSING and connect the SSD.")

manifest_path = VIDEO_PATH.with_suffix(".json")
if manifest_path.is_file():
    info = json.loads(manifest_path.read_text())
    print(f"Seed {info['seed']} | collision-free: {info['verified_success']} | "
          f"RMSE: {info['rmse_cm']:.2f} cm")
    print(f"Saved lap duration including warmup: {info['simulation_duration_s']:.1f} s")
print(f"Video: {VIDEO_PATH}")
```
